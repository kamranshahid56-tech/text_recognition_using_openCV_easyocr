```python
import easyocr
import cv2
import os
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "quest.jpeg"
img = cv2.imread(image_path)
reader = easyocr.Reader(["en"])
result = reader.readtext(image_path)
result
```

```python
for one in result:
    # Force coordinates to be integers
    top_left = tuple(map(int, one[0][0]))
    bottom_right = tuple(map(int, one[0][2]))

    text = one[1]

    # Now cv2.rectangle will receive (x, y) as integers
    img = cv2.rectangle(img, top_left, bottom_right, (255, 0, 0), 2)
    img = cv2.putText(img, text, top_left, cv2.FONT_HERSHEY_SIMPLEX, 1, (255, 0, 0), 2)

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)) # Convert BGR to RGB for correct colors in PLT
plt.show()
```
